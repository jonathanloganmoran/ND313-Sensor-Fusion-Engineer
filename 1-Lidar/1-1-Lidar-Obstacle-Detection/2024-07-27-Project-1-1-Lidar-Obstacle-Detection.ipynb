### Visualisation

#### `E1.5.0`: `renderPointCloud()`

For this task, our work is relatively simple. We want to call the Point Cloud Library (PCL) `render::renderPointCloud()` function from inside our "P1.1 programme loop", i.e., inside the `projectPipeline()` function we created above.

##### The `renderPointCloud()` function call

To display the "contents" of the current "frame" in our processing pipeline, i.e., the LiDAR point cloud scan, we use the function call to `render::renderPointCloud()`. This function (defined in the [`render.cpp`]() file) uses the Point Cloud Library (PCL) to render each "dot" (a 3D point measurement) onto the viewer. 

Optionally, we can provide a `Color` instance (defined in [`render.h`](https://github.com/jonathanloganmoran/ND313-Sensor-Fusion-Engineer/blob/main/1-Lidar/1-1-Lidar-Obstacle-Detection/src/render/render.h)) with RGB values to specify the colour we wish to visualise each point with. By default (i.e., with no `Color` input argument provided), the LiDAR points will be visualised using their _intensity_ attribute values. By not providing a `Color` variable as input, we assume the `inputCloudI` instance is, of course, a point cloud which contains `pcl::PointXYZI` instances, or, in other words, 3D  point values, _each_ with a corresponding _intensity_ measurement value.
